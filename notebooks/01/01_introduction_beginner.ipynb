### Load the iris dataset

Load iris into a pandas DataFrame and add the target column.

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["target"] = iris.target
df.head()
```

### Summary statistics

Print `.describe()` to see the range of each feature.

```python
# YOUR CODE HERE
df.describe()
```

### Feature distributions

Plot a histogram of each feature.

```python
import matplotlib.pyplot as plt

# YOUR CODE HERE
df.hist(figsize=(9, 6))
plt.tight_layout()
```

### A feature relationship

Scatter petal length against petal width, coloured by class.

```python
# YOUR CODE HERE
plt.figure()
plt.scatter(df[df.columns[2]], df[df.columns[3]], c=df["target"])
plt.xlabel(df.columns[2])
plt.ylabel(df.columns[3])
```

### Feature correlations

Compute the correlation matrix and show it as an image.

```python
# YOUR CODE HERE
corr = df.corr()
plt.matshow(corr)
plt.colorbar()
```
