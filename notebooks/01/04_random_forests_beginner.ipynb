### Load the breast-cancer dataset

Load it into a DataFrame with a `target` column.

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

bc = load_breast_cancer()
df = pd.DataFrame(bc.data, columns=bc.feature_names)
df["target"] = bc.target
df.head()
```

### Class balance

Show the number of samples in each class.

```python
import matplotlib.pyplot as plt

# YOUR CODE HERE
df["target"].value_counts().sort_index().plot.bar()
```

### Feature distributions

Plot histograms for the first six features.

```python
# YOUR CODE HERE
df.iloc[:, :6].hist(figsize=(10, 6))
plt.tight_layout()
```

### A feature relationship

Scatter the first two features, coloured by class.

```python
# YOUR CODE HERE
plt.figure()
plt.scatter(df.iloc[:, 0], df.iloc[:, 1], c=df["target"])
plt.xlabel(df.columns[0])
plt.ylabel(df.columns[1])
```

### Feature correlations

Show the correlation matrix as an image.

```python
# YOUR CODE HERE
plt.matshow(df.corr())
plt.colorbar()
```
