### Split the data

Make a 70/30 stratified train/test split of iris.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0, stratify=y
)
```

### Train a logistic regression

Fit `LogisticRegression` on the training set.

```python
from sklearn.linear_model import LogisticRegression

# YOUR CODE HERE
model = LogisticRegression(max_iter=1000).fit(X_train, y_train)
```

### Accuracy

Report the test accuracy of the model.

```python
# YOUR CODE HERE
print(model.score(X_test, y_test))
```

### Confusion matrix

Show the confusion matrix on the test set.

```python
from sklearn.metrics import ConfusionMatrixDisplay

# YOUR CODE HERE
ConfusionMatrixDisplay.from_estimator(model, X_test, y_test, colorbar=False)
```
