### Task 1 — Linear regression  
*(Provided — just run this cell)*

**Goal:** Split diabetes 70/30 and fit `LinearRegression`.

**Do this:**
1. Run the cell — it loads diabetes, splits it, fits `LinearRegression`, and stores the model as `model`.

**Look for:** The final line prints the training score, just to confirm the fit ran.

```python
# Provided -- just run this cell.
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X, y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
model = LinearRegression().fit(X_train, y_train)
model.score(X_train, y_train)
```
