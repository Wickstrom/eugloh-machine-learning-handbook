### Load the diabetes dataset

Load it into a DataFrame with a `target` column.

```python
from sklearn.datasets import load_diabetes
import pandas as pd

diab = load_diabetes()
df = pd.DataFrame(diab.data, columns=diab.feature_names)
df["target"] = diab.target
df.head()
```

### Target distribution

Plot a histogram of the target (disease progression).

```python
import matplotlib.pyplot as plt

# YOUR CODE HERE
df["target"].hist()
```

### Feature distributions

Plot a histogram of each feature.

```python
# YOUR CODE HERE
df.drop(columns="target").hist(figsize=(10, 7))
plt.tight_layout()
```

### Correlation with the target

Show each feature's correlation with the target as a bar chart.

```python
# YOUR CODE HERE
df.corr()["target"].drop("target").sort_values().plot.barh()
```

### Feature correlations

Show the full correlation matrix as an image.

```python
# YOUR CODE HERE
plt.matshow(df.corr())
plt.colorbar()
```
