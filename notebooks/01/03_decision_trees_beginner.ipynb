### Load the wine dataset

Load it into a DataFrame with a `target` column.

```python
from sklearn.datasets import load_wine
import pandas as pd

wine = load_wine()
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df["target"] = wine.target
df.head()
```

### Class balance

Show how many samples belong to each of the three classes.

```python
import matplotlib.pyplot as plt

# YOUR CODE HERE
df["target"].value_counts().sort_index().plot.bar()
```

### Feature distributions

Plot a histogram of each feature.

```python
# YOUR CODE HERE
df.drop(columns="target").hist(figsize=(11, 8))
plt.tight_layout()
```

### A feature relationship

Scatter alcohol against flavanoids, coloured by class.

```python
# YOUR CODE HERE
plt.figure()
plt.scatter(df["alcohol"], df["flavanoids"], c=df["target"])
plt.xlabel("alcohol")
plt.ylabel("flavanoids")
```

### Feature correlations

Show the correlation matrix as an image.

```python
# YOUR CODE HERE
plt.matshow(df.corr())
plt.colorbar()
```
