### Load two moons

Make the moons dataset and scale the features.

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler

X, y = make_moons(n_samples=400, noise=0.3, random_state=0)
Xs = StandardScaler().fit_transform(X)
```

### Plot the data

Scatter the two classes.

```python
import matplotlib.pyplot as plt

# YOUR CODE HERE
plt.figure()
plt.scatter(Xs[:, 0], Xs[:, 1], c=y)
plt.gca().set_aspect("equal")
```

### Class balance

Count the samples in each class.

```python
import numpy as np

# YOUR CODE HERE
print(np.bincount(y))
```

### Feature distributions

Plot a histogram of each feature.

```python
# YOUR CODE HERE
plt.hist(Xs[:, 0], alpha=0.6, label="x1")
plt.hist(Xs[:, 1], alpha=0.6, label="x2")
plt.legend()
```
