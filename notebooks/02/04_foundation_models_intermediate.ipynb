### Task 1 — Prepare the data  
*(Provided — just run this cell)*

**Goal:** Load digits, scale them, and split.

**Do this:**
1. Run the cell — it also defines a helper that selects the first n training examples of each class.

**Look for:** `first_n_per_class(5)` returns 50 indices.

```python
# Provided -- just run this cell.
import numpy as np
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

digits = load_digits()
X, y = digits.data / 16.0, digits.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0, stratify=y
)
scaler = StandardScaler().fit(X_train)
X_train, X_test = scaler.transform(X_train), scaler.transform(X_test)

def first_n_per_class(n):
    """Indices of the first n training examples of each class."""
    return np.concatenate([np.where(y_train == c)[0][:n] for c in np.unique(y_train)])

print(X_train.shape, X_test.shape)
```
